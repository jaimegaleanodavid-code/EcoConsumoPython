# Estructura de Datos: Pila (Stack) - LIFO

Una **Pila** es una estructura de datos lineal que sigue el principio **LIFO** (*Last In, First Out* - Último en entrar, Primero en salir). Es similar a una pila de platos: el último plato que colocas en la cima es el primero que retiras.

### Operaciones Principales
- **`push(elem)`**: Agrega un elemento en el tope de la pila.
- **`pop()`**: Retorna y elimina el elemento en el tope.
- **`peek()`**: Muestra el elemento en el tope sin eliminarlo.
- **`is_empty()`**: Comprueba si la pila está vacía.
- **`size()`**: Devuelve la cantidad total de elementos.
- **`clear()`**: Elimina todos los elementos de la pila.

```python
from collections import deque


class PilaAparatos:

    def __init__(self):

        self._elementos = deque()

    def push(self, elemento):
       
        self._elementos.append(elemento)

    def pop(self):
       
        if not self.is_empty():
            return self._elementos.pop()
        return None

    def peek(self):
      
        if not self.is_empty():
            return self._elementos[-1]
        return None

    def is_empty(self):
       
        return len(self._elementos) == 0

    def size(self):
        
        return len(self._elementos)

    def clear(self):
        
        self._elementos.clear()

    def obtener_lista(self):
       
        return list(self._elementos)
```

## Pruebas de Funcionamiento

A continuación, realizaremos pruebas simulando el registro de aparatos eléctricos para verificar que las operaciones respeten el orden de entrada y salida (LIFO).

```python

pila_prueba = PilaAparatos()


print(f"¿La pila está vacía?: {pila_prueba.is_empty()}")


pila_prueba.push(
    {"nombre": "Refrigerador", "potencia": 300, "horas": 24, "consumo": 216.0}
)
pila_prueba.push(
    {"nombre": "Televisor", "potencia": 120, "horas": 5, "consumo": 18.0}
)
pila_prueba.push(
    {"nombre": "Microondas", "potencia": 1000, "horas": 0.5, "consumo": 15.0}
)

print(f"Total de elementos tras 3 PUSH: {pila_prueba.size()}")


tope = pila_prueba.peek()
print(f"\nElemento en el tope (PEEK): {tope['nombre']}")


eliminado = pila_prueba.pop()
print(
    f"\nElemento desapilado (POP): {eliminado['nombre']} (Consumo: {eliminado['consumo']} kWh)"
)


nuevo_tope = pila_prueba.peek()
print(f"Nuevo elemento en el tope tras el POP: {nuevo_tope['nombre']}")


pila_prueba.clear()
print(
    f"\n¿La pila está vacía tras ejecutar clear()?: {pila_prueba.is_empty()}"
)
```
